```python
import pandas as pd
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.rc("font", family = "Malgun Gothic")
sns.set(font="Malgun Gothic", 
rc={"axes.unicode_minus":False}, style='white')

# # 한글 폰트 설정 (Mac)
# plt.rc('font', family='AppleGothic')
# # 마이너스 기호 깨짐 방지
# plt.rcParams['axes.unicode_minus'] = False

mt_fire = pd.read_excel('sanbul.xlsx', header=2)
mt_fire['fire'] = 1
mt_fire
```

```python
detail_reason = mt_fire['발생장소_시도'].value_counts().reset_index().head(10)
ax = sns.barplot(detail_reason,x='발생장소_시도', y='count', palette='seismic', errorbar=None)
ax.tick_params(axis='x', labelrotation=90) 
for container in ax.containers:
    ax.bar_label(container,  padding=3)
plt.xlabel('산불 발생 지역')

plt.ylabel('산불이 난 횟수')
plt.show()
```

```python
detail_reason = mt_fire['발생원인_세부원인'].value_counts().reset_index().head(10)
ax = sns.barplot(detail_reason,x='발생원인_세부원인', y='count', palette='seismic', errorbar=None)
ax.tick_params(axis='x', labelrotation=90) 
for container in ax.containers:
    ax.bar_label(container,  padding=3)
plt.xlabel('산불 발생 원인')

plt.ylabel('산불이 난 횟수')
plt.show()
```
