```python
import pandas as pd

df = pd.read_csv('./data1/머신러닝용_final.csv')
df
```

```python
# 년도 컬럼 신설
df['년도'] = df['일시'].str[:4].astype(int)
df
```

```python
df1 = pd.read_csv('전국_농지비율.csv',encoding='CP949')
df2 = pd.read_csv('전국_도시산림_인접지수.csv',encoding='CP949')

# 농지비율 컬럼 이름 통일
df_nongjy = df1.rename(columns={'발생일시_년':'년도','발생장소_시도':'시도'}) 
```

```python
# wui 컬럼 이름 통일
df_wui = df2.rename(columns={'시점':'년도','행정구역별':'시도'}) 
```

```python
# 산불-날씨 데이터와 merge
df_merge = df.merge(df_nongjy,on=['년도','시도']).merge(df_wui,on=['년도','시도'])
```

```python
df_merge.to_csv('최종전처리.csv', encoding='cp949', index=False)
```
